```python
#Import necessary libraries

import pandas as pd
import numpy as np
```

```python
#importing the statsmodels library
import statsmodels.api as sm
from sklearn import datasets
```

```python
#Reading the CSV file
df = pd.read_csv('https://raw.githubusercontent.com/oreoluwaodeyinka/IntroToMachineLearning/refs/heads/main/train.csv')
df.head(5)
```

```python
# Set the features
# Exclude 'ID' and 'medv' (target) from features
feature_columns = [col for col in df.columns if col not in ['ID', 'medv']]
setDf = df[feature_columns]

# Set the target
target = df[["medv"]]
```

## First Model

This model is an OLS models so it is attempting to tell us how one input variable, which in this case is "rm" which is average number of rooms per dwelling connects to and helps predict "medv" which is the median value of owner-occupied homes in $1000's

```python
#This is a section that attempts to create the first regression model
X = df["rm"]
y = target["medv"]

# Fit and make the predictions by the model
model = sm.OLS(y, X).fit()
predictions = model.predict(X)

# Print out the statistics
model.summary()
```

This model summary tells us a lot of things, most of which I don't quite understand yet, but from the reading I can gather a couple of things


*   R-squared: 0.904. R-squared will give you an estimate of the relationship between movements of a dependent variable based on an independent variable's movements.
*   Coefficient: 3.7004. This means that if the RM variable increases by 1, the predicted value of medv increases by 3.7004.



#Second Model

This model is a lasso models so it adds an absolute penalty to the model that shrinks weak feature coefficients down to exactly zero which will in essence "mute the noise" by not taking them into account and help us get a better understanding of how well the input variable "rm" connects to and helps predict "medv".

```python
from sklearn.linear_model import Lasso

# X contains your input feature (rm whch is average number of rooms)
# y contains your target values (medv which is median home value)

# alpha controls the penalty strength (higher alpha = more zeros)
lasso_model = Lasso(alpha=1.0)

# Ensure X is 2D for sklearn models
# If X is currently a pandas Series, convert it to a DataFrame (2D)
if isinstance(X, pd.Series):
    X = X.to_frame()

# Train the model
lasso_model.fit(X, y)

# Print R-squared score
print("Lasso R-squared:", lasso_model.score(X, y))

# Print coefficients
print("Lasso Coefficients:", lasso_model.coef_)


```

**This is the explanation gemini gave me on Lasso regression**

*   Lasso R-squared: The R-squared value of 0.4514 means that approximately 45.14% of the variance in the 'medv' (median house value) can be explained by the 'rm' feature in this Lasso model.

*   The output [6.9623752] is the coefficient for the 'rm' (average number of rooms) feature. This value indicates the impact of the 'rm' feature on the 'medv' (median house value) according to the Lasso model, with a positive coefficient suggesting that as the number of rooms increases, the median house value also tends to increase. So if the  RM variable increases by 1, the predicted value of medv increases by 6.9623752.




**What that explanation leads me to believe**

Ok so the R-squared of the OLS model was about double that of the lasso model which means that the OLS model shows more variance.
