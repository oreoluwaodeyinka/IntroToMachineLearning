# Data Cleaning and Data Pre - Processing

```
#This removes the cabin column from the dataset
train.drop('Cabin', axis=1, inplace= True)
```

```
#This function removes any null or missing values in the dataset
train.dropna(inplace=True)
```

The .getdummies function converts the unique values of a column into indicator variables. In this case there are only 2 possible values male or female so pandas will make 2 new columns indicating which category the person falls into by using inputting the value of the column as true or false.

The method drop_first = true is used to remove the first of the columns that it created above in order to prevent multiple columns basically giving the same information.

For the 2 lines of code below, it is converting the sex and embarked columns as outlined above and putting them in their own respective dataframes that will be used later.

```
sex = pd.get_dummies(train['Sex'], drop_first=True)
embark = pd.get_dummies(train['Embarked'], drop_first=True)
```

Since we've set the new columns up the way we want them we can go ahead and get rid of the columns we don't need anymore which are Sex and Embarked. I guess we also don't need name and Ticket anymore so we're using this opportunity to get rid of them as well.

```
train.drop(['Sex','Embarked','Name','Ticket'], axis=1, inplace= True)
```

Ok now we take a peek and see if the code worked by calling the .head() method so we can see glance at what columns and values are still left and yuh it worked so we're all set

```
train.head()
```

Now that we've checked, we can combine the three dataframes, train sex and embark into one dataframe called train that only has the data we want using pd.concat() which is a pandas fucntion used to glue together multiple series or dataframe objects.

```
train = pd.concat([train, sex, embark],axis=1)
```

And now we take another peek and see if the code worked by calling the .head() method so we can see glance at what columns and values are and yuh it worked so we're all set

```
train.head()
```

# Model building

```
#imorting the function that we will use for model building.
from sklearn.model_selection import train_test_split
```

The line of code below is splitting the code into training and testing sets


*   X-train: the subset of data that will be used to train the model
*   X_test: the subset of the data that will be used to test the model and make predicitons
*   y-train: the subset of data that will be used to train the model
*   y_test: the subset of the data that will be used to test the model and make predicitons

Ok so the x since it is the independent variable, is what we already know and have measured and that's gonna be all the variables/columns in the dataset that we are not trying to predict. We are trying to predict who survived and who didn't so we drop the survived column and feed the rest of the information to the model. This is the x: train.drop('Survived',axis=1)

The y is the dependent variable because based on what's going on with the other columns we make a prediction on whether or not the person survived which is y. This is the y: train['Survived']

test_size tells us how much of the data should be set aside for testing which in this case is 30% or 0.3

random_state i believe is just used to reference this exact process so if we need to recreate this we can tell it to call random state 101 and it can find it again.




```
X_train, X_test, y_train, y_test = train_test_split(train.drop('Survived',axis=1), train['Survived'],test_size=0.3, random_state=101)
```

```
#imorting the function that we will use for logisticregression.
from sklearn.linear_model import LogisticRegression
```

```
#This calls the LogisticRegression() function and stores an instance of it in logmodel
#changed this to increase max_iter from 100 to 1000 or higher to fix the error that was being shown below
logmodel = LogisticRegression(max_iter=1000)
```

```
#This is throwing a user error so gonna trouble shoot to see how to fix it.
#fixed it by increasing max_iter to 1000
#This is training the model
logmodel.fit(X_train, y_train)
```

```
#predict is storing is the the trained model's output and
predict = logmodel.predict(X_test)
```

## Classification report

```
#imorting the function that we will use for creating a report to understand how well the model is performing.
from sklearn.metrics import classification_report
```

```
#printing out the report
print(classification_report(y_test,predict))
```

Ok so looking at this breakdown, we can see a couple things:

First I want to pount out there is a 0 and a 1 column in our report. I believe the 0 column represents False meaning the person did not survive and the 1 represents true so the person did survive.

For the instances were the person did not survive aka the 0 column
*   precision which is when the model makes a prediction and is actually right is at 0.82 so about 82% of the time when the person didn't survive, it predicted that they didn't survive.
*   recall was 0.85 which means out of all the people who actually did not survive the model was able to find them about 85% of the time.
* f1-score is the average of recall and precision so 0.84
* support is the number of occurences so the model predicted that 128 people did not survive.


For the instances were the person did survive aka the 1 column

* precision which is when the model makes a prediction and is actually right is at 0.77 so about 77% of the time when the person did survive, it predicted that they did survive.
*   recall was 0.72 which means out of all the people who actually did  survive the model was able to find them about 77% of the time.
* f1-score is the average of recall and precision so 0.74
* support is the number of occurences so the model predicted that 86 people did survive.


Accuracy, macro averages, and weighted averages are summary stats so they tell us overall how the model is doing and it seems to be doing pretty well.




## Confusion Matrix

```
#importing the confusion matrix function and calling it on the model's outputs.
from sklearn.metrics import confusion_matrix
#y_test is the actual real output and predict is what the model outputted.
print(confusion_matrix(y_test, predict))
```

The confusion matrix is telling us this:
* 109 - True Negative
* 24 - False negative
* 19 - False Positives
* 62 - True Posiitves

SO it's giving us the context that the model identified 109 true negatives and 62 true positives which is very high since we know the totsl number possible for everything is 214.

```
# This section of code creates a heatmap based on the confusion matrix we created above

plt.figure(figsize=(9,9))
sns.heatmap(confusion_matrix(y_test, predict), annot=True, fmt=".3f", linewidths=.5, square = True, cmap = 'Blues_r');
plt.ylabel('Actual label');
plt.xlabel('Predicted label');
all_sample_title = 'Accuracy Score: {0}'.format(logmodel.score(X_test, y_test))
plt.title(all_sample_title, size = 15);
```
